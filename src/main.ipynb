```python
from data_processing import DataProcessing
from a_priori import APriori

from pyspark.sql import functions as F, SparkSession
from pyspark.sql.types import ArrayType, IntegerType
from pyspark.sql import DataFrame
from pathlib import Path
```

```python
# Start spark
spark = SparkSession.builder.appName("main").getOrCreate()

ROOT = Path().resolve().parent
data_path = ROOT / "Data" / "sales.dat"
# Load the dataframe at ROOT/Data/sales.dat by default
df = DataProcessing(filepath=data_path).load_dataframe(spark=spark)

priori = APriori(df=df, s=10)
```

```python
chars_with_support = priori._first_filter()
chars = chars_with_support.select(F.col('set'))

permutes = priori.permute_increasing(chars)
```
